```python
#list
#ist is an ordered, mutable collection.
#duplicate allowed

list_data = ["Ravi","24","Java","Java"]
print(list_data)
print(list_data[0])
```

```python
#tupple
#tuple is an ordered, immutable collection.
#duplicate allowed

tupple_data =("Ravi","24","Java","Java")
print(tupple_data)
print(tupple_data[0])
print(tupple_data[1])
```

```python
#set
#set stores unique elements.
#duplicate not allowed

set_data = {"Ravi","24","Java","Java","java"}
print(set_data)

#print(set_data[0])--# we can't do it's unordered

#need to convert into list then can access the data
list_data = list(set_data)
print(list_data)
print(list_data[0])
```

```python
#dictionary
#dictionary stores key-value pairs.
dict_data = {"name":"Ravi","age":"24","language":"Java"}
print(dict_data)
print(dict_data["name"])
```
