```python
#withput list comprehension

square_data = []

#for i in range(5):  # means 0 to 4
for i in range(1,5):  # means 1 to 4    
    square_data.append(i**2)

print(square_data)   


#with list comprehension

result_square = [i**2 for i in range(1,5)]
print(result_square)
```

```python
# for even data

result_even = [i**2 for i in range(1,5) if i%2==0]
print(result_even)
```

```python
# for odd data

result_odd = [i**2 for i in range(1,5) if i%2!=0]
print(result_odd)
```

```python
#with else if 

result = ["Even" if i%2==0 else "Odd" for i in range(1,5)]
print(result)
```

```python
#dictionay comprehension

dict_data = {i:i**2 for i in range(1,5)}
print(dict_data)
```

```python
#dictionay comprehension with if 

dict_data = {i:i**2 for i in range(1,5) if i%2==0 }
print(dict_data)
```

```python
#set comprehension

set_data = {i**2 for i in range(1,5)}
print(set_data)
```
